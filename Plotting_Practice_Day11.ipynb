# Plotting in Python

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

### Relative Permeability Plots

```python
sw= np.linspace(0,1,100)
```

```python
sw
```

```python
krw = sw**3

kro = (1-sw)**2
```

```python
krw
```

```python
kro
```

## Line plot for Relative permeability

```python
plt.plot(sw,krw)
plt.plot(sw,kro)
```

```python
plt.style.use("classic")

plt.figure(figsize=(8,6))

plt.plot(sw,krw, label = "Water Relative Permeability",color="darkblue",lw=0.7)  #lw = linewidth
plt.plot(sw,kro, label = "Oil Relative Permeability",color="brown",lw=5)

plt.xlabel("Water Saturation (fraction)")

plt.ylabel("Relative Permeability (fraction)")

plt.legend()
plt.title("Relative Permeability Plots")
plt.grid()
```

## Histograms: Distribution: how many times a number is occuring

```python
plt.hist(df["BORE_OIL_VOL"])
```

```python
plt.hist(df["AVG_CHOKE_SIZE_P"])
```

```python
plt.figure(figsize=(8,4))
plt.hist(df["BORE_OIL_VOL"],label="Oil Production",color="red", bins = 50)
plt.hist(df["BORE_WAT_VOL"],label="Water Production",color="blue",alpha=0.5, bins= 50)
plt.grid(True)
plt.legend()
```

### Bins = How many intervals the data would be divided
Less no of Bins = More coarse and less detailed dataset
More no of Bins = More Thin and more detailed dataset

Sturge rule
No of Bins = Log2(n) + 1  (where n is no of data points)

### Alpha = controls transparency in case of overlapping histograms

Alpha = 1 (Fully transparent)
Aplha = 0 (Fully Opaque)

## PIE CHARTS

```python
df["well open"].value_counts().index
```

```python
plt.pie(df["well open"].value_counts(),labels = df["well open"].value_counts().index,autopct='%1.2f%%')

#labels = index of df["well open"] values which is full day partial and shut in
#autopct = '%1.2f%' it shows the values in % terms
```
