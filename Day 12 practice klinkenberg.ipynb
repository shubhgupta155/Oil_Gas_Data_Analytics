## Creating plot for Klinkenberg and asking user values

```python
def klinkenberg(kg,pm,k):
    
    """
    This klinkenberg function calculates absolute permeability for given gas permeability at a given mean pressure.
    Plots included
    Asking user defined values
    
    INPUTS:
    kg=> gas permeability in lab at a given mean pressure pm
    pm => mean pressure
    k=>User initial guess for absolute perm.
    
    return:
    k=> value of absolute permeability 
    
    
    
    """
    
    pm_user = float(input("Enter the value of mean pressure(psi): "))
    count = 0
    print(f"The value of kl_{count} is {k}")

    while (abs(6.9*(k**0.64)+pm*k-pm*kg)>0.000000000001):
        print("=======================")
        updated_value = (6.9*(k**0.64)+pm*k-pm*kg)/(4.416*(k**(-0.36)) + pm) #f(k_l)/f'(k_l)
        k = k - updated_value
        count = count+1
        print(f"The value of kl_{count} is {k}")


    print(f"The final value of Perm K is : {k}")
    
    m = (kg-k)/((1/pm)-0)
    
    x_axis = np.linspace(0,1,200)
    y_axis = m*x_axis+k
    
    
    kg_user = m*(1/pm_user)+k
    
    print(f"The value of Kg is : {kg_user:.3f} mD at mean pressure of {pm_user} psi ")
    
    plt.style.use("classic")
    plt.figure(figsize=(10,8))
    
    plt.plot(x_axis,y_axis, label= "Klinkenberg curve")
    plt.scatter(1/pm, kg, s=10, c = "red", label = "Lab condition reading")
    plt.scatter(1/pm_user, kg_user, s=15, c = "green", label = "User defined reading")
    
    plt.axhline(k, color = "black", lw = 4, label = f"absolute permeability: {k:.3f} mD")  #For drawing horizontal axis line
    
    plt.ylim(0,100)  #Limiting values of y and x using plt.ylim
    plt.xlim(0,1)
    
    plt.xlabel = ("1/Mean pressure")
    plt.ylabel = ("Gas Permeability Kg")
    plt.legend()
    plt.grid(True)
    plt.show()
    
    return k
```

```python
klinkenberg(35,2,56)
```
