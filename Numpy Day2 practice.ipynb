## Overlay a Smooth Distribution Curve (PDF)

### For normal distributions, we use the Probability Density Function (PDF) to plot a smooth curve.

```python
from scipy.stats import norm

x = np.linspace(min(data), max(data), 100)  # Generate 100 evenly spaced values from min to max
pdf = norm.pdf(x, loc=50, scale=10)  # Compute the probability density function for given x values

plt.plot(x, pdf, 'r', linewidth=2)  # Plot the smooth normal distribution curve
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Step 1: Generate Normal Distribution Data
data = np.random.normal(loc=50, scale=10, size=10000)  # Mean=50, Std Dev=10

# Step 2: Plot the Histogram
plt.hist(data, bins=50, density=True, alpha=0.6, color='b')

# Step 3: Generate and Plot Normal Distribution Curve
x = np.linspace(min(data), max(data), 100)
pdf = norm.pdf(x, loc=50, scale=10)
plt.plot(x, pdf, 'r', linewidth=2)

# Step 4: Add Labels and Title
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution Curve")

# Step 5: Show the Plot
plt.show()
```
